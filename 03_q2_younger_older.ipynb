```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
clients_demo = pd.read_csv("02_cleaned_data_files/cleaned_clients_demographics.csv")
clients_var = pd.read_csv("02_cleaned_data_files/cleaned_clients_variation.csv")
web_events = pd.read_csv("02_cleaned_data_files/cleaned_web_events.csv")
```

```python
# Select all clients who used the online process
online_clients = clients_demo[
    clients_demo['client_id'].isin(web_events['client_id'])
].drop_duplicates(subset='client_id')

# Analyze their age
online_clients['clnt_age'].describe().round(1)
```

```python
# Create age groups
online_clients['age_group'] = pd.cut(
    online_clients['clnt_age'],
    bins=[0, 35, 50, 65, float('inf')],
    labels=['Under 35', '35-49', '50-64', '65+'],
    right=False
)

# Calculate percentage of clients in each group
age_distribution = (
    online_clients['age_group']
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(1)
)

print(age_distribution)
```

```python
import matplotlib.pyplot as plt

ax = age_distribution.plot(kind='bar', figsize=(12, 6))

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=5, fontsize=13, fontweight='bold')

plt.title('Age Distribution of Vanguard Online Clients', fontsize=17, pad=20)
plt.xlabel('Age Group', fontsize=13)
plt.ylabel('Percentage of Clients (%)', fontsize=13)

plt.xticks(rotation=0, fontsize=12)
plt.yticks(fontsize=12)

plt.ylim(0, 38)
plt.tight_layout()
plt.show()
```

##### Vanguard's online clients are predominantly middle-aged, with 57.9% aged 35–64 and an average age of 46.4. The largest segment is 50–64 (31.9%). This suggests that the digital experience should prioritize intuitive navigation and clear instructions across age groups.
